## 3. لماذا نستخدم NumPy؟

قوائم Python العادية مناسبة لكثير من المهام، لكن العمليات العددية على كميات كبيرة من البيانات تكون غالباً أكثر كفاءة باستخدام مصفوفات NumPy.

مصفوفة NumPy (`ndarray`) تخزّن عناصر متجانسة النوع غالباً، وتسمح بتنفيذ العملية على مجموعة عناصر دفعة واحدة. هذا الأسلوب يسمى **Vectorization**، وهو يقلل الحاجة إلى كتابة حلقات `for` في كثير من العمليات العددية.

NumPy أيضاً أساس مهم لمكتبات مثل Pandas وScikit-learn.

````python
import numpy as np

# إنشاء مصفوفة من قائمة
a = np.array([1, 2, 3, 4])

# عمليات على جميع العناصر دون كتابة for
print(a * 2)       # [2 4 6 8]
print(a + a)       # [2 4 6 8]

# دوال إحصائية أساسية
print(a.mean())    # المتوسط
print(a.std())     # الانحراف المعياري
print(a.sum())     # المجموع
````

## 4. شكل المصفوفة ونوع البيانات

خصائص مهمة لأي مصفوفة:

- `shape`: عدد الصفوف والأعمدة في كل بُعد.
- `dtype`: نوع العناصر.
- `ndim`: عدد الأبعاد.

في المثال التالي لدينا مصفوفة ثنائية الأبعاد فيها صفّان وثلاثة أعمدة.

````python
m = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("shape:", m.shape)  # (2, 3)
print("dtype:", m.dtype)  # نوع العناصر
print("ndim:", m.ndim)    # 2
````

## 5. إنشاء مصفوفات جاهزة

توفر NumPy دوالاً لإنشاء مصفوفات شائعة دون كتابة كل عنصر يدوياً:

| الدالة | الاستخدام |
|---|---|
| `np.zeros((2, 3))` | مصفوفة من الأصفار |
| `np.ones(5)` | خمسة عناصر قيمتها 1 |
| `np.arange(0, 10, 2)` | قيم تبدأ من 0 وتتقدم بمقدار 2 |
| `np.linspace(0, 1, 5)` | خمس قيم متساوية التباعد بين 0 و1 |
| `np.random.rand(3, 3)` | مصفوفة عشوائية 3×3 بين 0 و1 |

````python
print(np.zeros((2, 3)))
print(np.ones(5))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.random.rand(3, 3))
````

## 6. الفهرسة والتقطيع — Indexing & Slicing

الفهرسة في Python تبدأ من الصفر:

- `m[0, 1]`: العنصر في الصف الأول والعمود الثاني.
- `m[:, 1]`: جميع الصفوف من العمود الثاني.
- `m[1, :]`: جميع أعمدة الصف الثاني.
- `a[1:3]`: العناصر من الفهرس 1 إلى ما قبل الفهرس 3.

تذكّر أن نهاية التقطيع غير مشمولة.

````python
print(m[0, 1])  # 2
print(m[:, 1])  # [2 5]
print(m[1, :])  # [4 5 6]
print(a[1:3])   # [2 3]
````

## 7. الفلترة باستخدام Boolean Mask

يمكننا إنشاء شرط ينتج قيماً منطقية `True` و`False`، ثم استخدامه لاختيار العناصر التي تحقق الشرط.

هذه الفكرة ستعود معنا كثيراً عند استخدام Pandas.

````python
values = np.array([5, 12, 7, 20])

mask = values > 6
print(mask)            # [False True True True]
print(values[mask])    # [12  7 20]
print(values[values % 2 == 0])  # [12 20]
````

## 8. فهم المحور — axis

هذه من النقاط التي تربك المبتدئين:

- `axis=0`: نجمع على امتداد الصفوف، فتكون النتيجة قيمة لكل عمود.
- `axis=1`: نجمع على امتداد الأعمدة، فتكون النتيجة قيمة لكل صف.

باستخدام المصفوفة `m` التي أنشأناها سابقاً، مجموع العمود الأول هو 1 + 4 = 5، ومجموع الصف الأول هو 1 + 2 + 3 = 6.

````python
print(m.sum(axis=0))  # [5 7 9] مجموع كل عمود
print(m.sum(axis=1))  # [6 15] مجموع كل صف
````

## 9. ملاحظة مهمة: النسخ والمراجع

عندما تكتب `b = a` فأنت لا تنشئ مصفوفة مستقلة؛ بل تجعل المتغيرين يشيران إلى البيانات نفسها. لذلك قد يؤثر تعديل أحدهما في الآخر.

استخدم `copy()` عندما تحتاج نسخة مستقلة.

````python
original = np.array([1, 2, 3])
reference = original
independent_copy = original.copy()

reference[0] = 99

print("original:", original)                 # يتغير
print("independent copy:", independent_copy) # لا يتغير
````
