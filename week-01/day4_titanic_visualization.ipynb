# Day 04 — Exploratory Data Analysis & Data Visualization

This notebook explores the Titanic dataset visually before any modeling. It creates an age histogram with KDE, a fare-by-class boxplot, and a correlation heatmap for numeric features. All figures are saved to `images/` for reuse in the repository journal.


```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Reproducible output directory
OUTPUT_DIR = Path("images")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Load the original Titanic dataset
df = sns.load_dataset("titanic")
print(f"Dataset shape: {df.shape}")
display(df.head())
```

## 1. Passenger Age Distribution

A histogram reveals the concentration and skew of passenger ages, while the KDE curve provides a smoothed view of the distribution. The figure helps assess whether a mean-based fill would be sensitive to the long tail.


```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x="age", bins=20, kde=True, color="mediumpurple")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "age_distribution.png", dpi=160, bbox_inches="tight")
plt.show()
```

## 2. Fare Distribution by Passenger Class

A boxplot compares the fare distributions across passenger classes. The median, interquartile range, whiskers, and individual outliers make differences in ticket prices visible.


```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="pclass", y="fare", palette="pastel")
plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Fare")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "fare_by_class_boxplot.png", dpi=160, bbox_inches="tight")
plt.show()
```

## 3. Numerical Feature Correlation

The Pearson correlation matrix summarizes pairwise linear associations between numeric columns. Correlation describes association, not causation; missing values are handled pairwise by Pandas.


```python
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr(numeric_only=True)

plt.figure(figsize=(12, 9))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
)
plt.title("Titanic Numerical Feature Correlation")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "correlation_heatmap.png", dpi=160, bbox_inches="tight")
plt.show()

print("Saved visualizations:")
for image_path in sorted(OUTPUT_DIR.glob("*.png")):
    print(image_path)
```

## Key Takeaways

- Passenger ages are concentrated in young adulthood, with a right tail toward older ages.
- First-class fares show a much wider spread and more extreme high-value observations than second- and third-class fares.
- Passenger class and fare are strongly negatively correlated (approximately -0.55); because class 1 is the highest class and class 3 the lowest, the numeric coding runs opposite to fare level.
- Passenger class and survival are also negatively correlated (approximately -0.34), indicating higher survival among lower numerical class codes (the more privileged classes). This is an association, not proof of causation.
