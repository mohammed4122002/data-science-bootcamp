```python
import pandas as pd
import seaborn as sns

# تحميل البيانات وفحصها
df = sns.load_dataset("titanic")
print("--- Shape ---")
print(df.shape)
print("\n--- Missing Values Before Cleaning ---")
print(df.isna().sum())

# معالجة القيم الناقصة والصفحات
df["age"] = df["age"].fillna(df["age"].median())  # تعبئة العمر بالوسيط لمقاومة الشواذ
df = df.drop(columns=["deck"])                    # حذف عمود deck لارتفاع نسبة الفاقد
df = df.dropna(subset=["embarked"])               # حذف الصفوف ذات القيم الناقصة في الميناء
df = df.drop_duplicates()                         # إزالة التكرارات

print("\n--- Missing Values After Cleaning ---")
print(df.isna().sum())
```
